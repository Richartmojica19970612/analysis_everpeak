```python
# Convertir las columnas de traffic a tipo fecha con pd.to_datetime()
traffic['update_time_utc'] = pd.to_datetime(traffic['update_time_utc'], errors='coerce')
traffic['update_time_utc_week_ago'] = pd.to_datetime(traffic['update_time_utc_week_ago'], errors='coerce')
# verificar el cambio
traffic.info()
```

```python
# Limpia separadores y convierte columnas numéricas en eco
eco['city_gdp_capita'] = (
    eco['city_gdp_capita']
    .astype(str)                         # asegura que todo sea string
    .str.replace('.', '', regex=False)   # elimina separadores de miles
    .str.replace(',', '.', regex=False)  # convierte comas en puntos decimales
    .astype(float)                       # convierte a float
)

# unemployment_pct: quitar % y convertir a float
eco['unemployment_pct'] = (
    eco['unemployment_pct']
    .astype(str)
    .str.replace('%', '', regex=False)   # elimina símbolo de porcentaje
    .str.replace(',', '.', regex=False)  # convierte comas en puntos decimales
    .astype(float)
)

# population_m: convertir a float
eco['population_m'] = (
    eco['population_m']
    .astype(str)
    .str.replace(',', '.', regex=False)  # convierte comas en puntos decimales
    .astype(float)
)

# Crear nueva columna population en unidades absolutas
eco['population'] = eco['population_m'] * 1_000_000

# Verificar cambios
eco.info()
eco.head(3)
```

- Calcula el promedio **solo de las métricas de tráfico más relevantes**: como `jams_delay`, `traffic_index_live`, `jams_length_kms`, `jams_count`, `mins_delay`, y tiempos de viaje (`travel_time_live_per_10kms_mins` y `travel_time_hist_per_10kms_mins`).

<details>
<summary>Haz clic para ver la pista</summary>

```python
# Calcular los promedios de tráfico por ciudad, país y año
traffic_city_year_2024 = (
    traffic_2024
    .groupby(['city', 'country', 'year'], as_index=False)
    .agg({
        'jams_delay': 'mean',
        'traffic_index_live': 'mean',
        'jams_length_in_kms': 'mean',
        'jams_count': 'mean',
        'mins_delay': 'mean',
        'travel_time_live_per_10kms_mins': 'mean',
        'travel_time_hist_per_10kms_mins': 'mean'
    })
)

# Mostrar resultado
traffic_city_year_2024.head()
```

```python
# Crear histograma para ver la distribución de la economía (city_gdp_capita)
plt.figure(figsize=(8,6))
sns.histplot(merged['city_gdp_capita'], bins=15, kde=True, color="green")

plt.title("Distribución del PIB per cápita de las ciudades", fontsize=14)
plt.xlabel("PIB per cápita (USD)")
plt.ylabel("Frecuencia")
plt.show()
```
